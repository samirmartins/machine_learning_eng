# Assignment 1 (T1) — Analysis of Real Data in Python

## 1. Purpose of the assignment

- Consolidate, in a practical and integrated way, the initial contents of the course, especially items **1 to 4** of the syllabus:
- Prepare the group for the second assignment of the course
- Get used to essential tools in Machine Learning
- Develop a complete computational study, in Python, using **real data** and producing a clear, organized, reproducible, and technically well-justified analysis.

---

## 2. Assignment topic

Each group shall choose **a problem of interest** and investigate it from a **real database**, obtained from a **reliable source**, as discussed throughout the course.

**IMPORTANT:** the topic to be addressed must be agreed with the professor, through proposals presented by the group.

The purpose of the assignment is not, at this moment, to build an advanced predictive or classificatory modeling project, but to show that the group is able to:

- obtain real data;
- organize and manipulate these data in Python;
- perform exploratory and descriptive statistical analysis;
- produce informative visualizations;
- discuss the problem with language appropriate to the course;
- relate the analyzed database to a possible machine learning task (topic for Assignment 2).

---

## 3. Central requirements: use of real data and exploratory analyses

The assignment must obligatorily use **real data**.

These data must:

- originate from **reliable sources**;
- have their source clearly indicated;
- be adequate to the scope of the course;
- allow a minimally relevant analysis.

Possible sources will be discussed and exemplified in the course.



---

## 4. What the assignment must contain

The report (in .ipynb format) must cover, in an organized way, the following elements.

### 4.1. Identification
- title of the assignment;
- names of the group members;
- link to the group's GitHub repository.

### 4.2. Problem definition
- objective description of the topic;
- question, problem, or motivation of the study;
- brief justification of the relevance of the topic.

### 4.3. Source and acquisition of the data
- the format of data acquisition (file, API, data portal, etc.);
- explanation of how the data were obtained;
- remarks on reliability, context, and limitations.
- a justification of the choice of the database and of the data for the assignment.

### 4.4. Reading and organizing the database in Python
- loading of the database;
- description of the variables;
- initial structuring of the data;
- remarks on format, types, and consistency.

### 4.5. Data treatment and preparation
- cleaning, when necessary;
- treatment of missing values, inconsistencies, or duplicates;
- relevant standardizations and transformations;
- justification of the choices made.

### 4.6. Exploratory and statistical analysis
The group must use, when it makes sense for the chosen database:

- measures of central tendency;
- descriptive analysis;
- distributions;
- relationships between variables;
- discussion of correlation;
- comments on causality and spurious correlations, when applicable;
- situations involving dependence, independence, or conditional probability, when appropriate.

### 4.7. Visualizations
The notebook must present a minimum of 6 distinct graphs and visualizations, useful for understanding the data, with titles, legends, and adequate interpretation.

**IMPORTANT:** all constructed graphs must be analyzed, and not only presented.

### 4.8. Conceptual closing linked to machine learning
At the end, the group must include an introductory discussion answering, whenever possible:

- which type of machine learning problem could be associated with the analyzed database;
- whether the problem is closer to classification, regression, or another formulation;
- which difficulties could arise in a later modeling stage;
- how the concepts of overfitting, underfitting, and bias-variance could appear in this context.

**IMPORTANT:** It is not mandatory for the group to implement, in this T1, complete machine learning models. The focus is the conceptual and analytical preparation. It is recommended that the notebook alternate between code cells and **markdown** cells, for better organization of the work.

---

## 5. Mandatory deliverables

Each group must deliver:

1. **a report in an `.ipynb` file**;
2. **a Conda environment file (`.yml`)** of the group;
3. **the link to the GitHub repository**, informed inside the notebook, containing the group's presentation.
4. **The data that were used in the assignment (`.csv` or other format)**

---

## 6. Mandatory organization on GitHub

The assignment must be organized in a repository on **GitHub**.

The repository must be clean, understandable, and adequately structured, containing at least:

- the main notebook of the assignment;
- the `.yml` environment file;
- possible auxiliary files;
- data and instructions for data access, when applicable;
- presentation material;
- README.md file, containing the description of the repository.

The idea is that, throughout the course, it is possible to constitute **a course repository with all the developed assignments**.

---

## 7. Identification of each member's contribution

It is mandatory to indicate, in a visible way, **what was done by each group member**.

This identification must be done **by colors** throughout the notebook.

Example of convention:

- **Blue**: Student 1
- **Green**: Student 2
- **Red**: Student 3
- **Another color**: Student 4, if any

This convention must appear right at the beginning of the notebook.

In addition:

- each relevant excerpt must allow identifying who produced it;
- the division of tasks must be clear;
- all members must contribute and all must present.

- The presentation of the assignment must last **up to 18 minutes**. It must be objective, clear, and organized, and will be followed by **5 minutes** of questions.
- I suggest an average of 1 to 1.5 slides per minute. Therefore, for an 18-minute presentation, from 18 to 27 slides.
- Presentations **over 18 minutes** will be interrupted. In these cases, the group will have the presentation grade impaired.
