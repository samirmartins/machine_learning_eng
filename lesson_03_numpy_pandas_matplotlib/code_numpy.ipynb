# NumPy: arrays, vectorized operations, and matrices

This notebook brings together short, executable examples of:
- creating arrays and numerical sequences;
- inspecting `shape`, `dtype`, and `size`;
- indexing and slicing;
- vectorized operations and useful functions;
- matrices and solving linear systems.


```python
import numpy as np
```

## 1. Creating arrays and sequences

```python
grades = np.array([6.5, 7.0, 8.5, 9.0])
time = np.linspace(0, 10, 6)
evens = np.arange(0, 12, 2)

print("grades:", grades)
print("time:", time)
print("evens:", evens)
```

## 2. Shape, dtype, and size

```python
measurements = np.array([[1.2, 1.5, 1.7],
                    [2.0, 2.2, 2.4]])

print("shape:", measurements.shape)
print("dtype:", measurements.dtype)
print("size:", measurements.size)
```

## 3. Indexing and slicing

```python
M = np.array([[10, 11, 12], 
              [20, 21, 22], 
              [30, 31, 32]])


#print("element [0, 1]:", M[0, 1])
#print("third column:", M[:, 2])
#print("submatrix:\n", M[1:, :2])

submatrix = M[1:, 1:(M.shape[1])]
print(submatrix)
```

## 4. Vectorized operations

```python
temps_c = np.array([18, 21, 25, 27])

temps_f = temps_c * 9 / 5 + 32

increased_temps = temps_c + 2

print("temperatures in Celsius:", temps_c)
print("temperatures in Fahrenheit:", temps_f)
print("temperatures + 2:", increased_temps)
```

## 5. Useful functions

```python
data = np.array([4, 6, 8, 10, 12])

print("sum:", np.sum(data))
print("mean:", np.mean(data))
print("max:", np.max(data))
print("min:", np.min(data))
```

## 6. Matrices and linear algebra

```python
A = np.array([[3, 1],
              [1, 2]])
B = np.array([[2, 0],
              [1, 4]])
b = np.array([9, 8])

print("A @ B =\n", A @ B)
print("solution of A x = b:", np.linalg.solve(A, b))
```

## 7. Integrated example

We will generate a signal, calculate simple statistics, and reshape the data into a matrix.

```python
x = np.linspace(0, 2 * np.pi, 8)
signal = np.sin(x)
signal_matrix = signal.reshape(2, 4)

print("signal:", np.round(signal, 3))
print("signal mean:", np.mean(signal))
print("2x4 matrix:\n", np.round(signal_matrix, 3))
```
