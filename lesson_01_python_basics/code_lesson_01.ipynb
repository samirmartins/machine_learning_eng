```python
# Imports
import matplotlib.pyplot as plt

# Comment: Python features example

# Variables
name = "Ana"
age = 22

# Conditional
if age >= 18:
    print(f"{name} is an adult.")

# List and loop
grades = [7.5, 8.0, 9.2]

for i, grade in enumerate(grades):
    print(f"Grade {i+1}: {grade}")

# Function
def mean(values):
    return sum(values) / len(values)

print("Mean:", mean(grades))

# Simple plot
plt.plot([1, 2, 3], [7.5, 8.0, 9.2], marker='o')
plt.title("Grades")
plt.xlabel("Exam")
plt.ylabel("Grade")
plt.grid(True)
plt.show()
```
